## Introduction

Quarto can render `.ipynb` notebooks **directly** — no conversion step needed.
Just place your notebook in a `posts/` subfolder and it becomes a blog post.

All code cells run and their outputs are embedded in the final article.

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate data
x = np.linspace(0, 2 * np.pi, 300)
y1 = np.sin(x)
y2 = np.sin(x + np.pi/4)

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(x, y1, label='sin(x)', linewidth=2)
ax.plot(x, y2, label='sin(x + π/4)', linewidth=2, linestyle='--')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.legend()
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
```

## Interactive Plotly Output

Plotly figures render interactively when the notebook is converted:

```python
import plotly.express as px
import pandas as pd

df = pd.DataFrame({
    'x': np.random.randn(200),
    'y': np.random.randn(200),
    'category': np.random.choice(['A', 'B', 'C'], 200)
})

fig = px.scatter(
    df, x='x', y='y', color='category',
    template='simple_white',
    height=400
)
fig.show()
```

## DataFrames as Tables

Pandas DataFrames render as clean HTML tables:

```python
import pandas as pd

results = pd.DataFrame({
    'Model':      ['Transformer', 'LSTM', 'CNN', 'MLP'],
    'Accuracy':   [0.943, 0.912, 0.898, 0.871],
    'Parameters': ['125M', '45M', '12M', '3M'],
    'Training hrs': [24, 18, 6, 2]
})

results.style.format({'Accuracy': '{:.1%}'})
```

## Converting Your Notebooks

To publish any notebook as a post:

1. Copy your `.ipynb` to `posts/your-post-name/`
2. Add YAML front matter in a `raw` cell at the top
3. Run `quarto render` — it just works

Or use the conversion script: `./scripts/nb2post.sh my_notebook.ipynb "My Post Title"`
